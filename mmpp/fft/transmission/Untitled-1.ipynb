# ⚠️WAŻNE: Różnice w slicingu między dyspersją a transmisją

## 🔍 Analiza problemu

### ✅ Dyspersja - WSPIERA slicing
```python
# TO DZIAŁA! ✅
disp_result = job[0].m_layer[:100,...].fft.dispersion.compute_1d(axis="x")
```

**Dlaczego działa:**
- `FFTDispersionInterface` otrzymuje `slice_info` (`:100`) w konstruktorze
- `SpinWaveAnalyzer` wykorzystuje `slice_info` do inteligentnego ładowania tylko potrzebnych danych
- Slicing jest przetwarzany przez `_normalize_slice()` i `_indexer_for_tmax()`

### ❌ Transmisja - NIE WSPIERA slicingu (jeszcze)
```python
# TO NIE DZIAŁA POPRAWNIE! ❌
trans_result = job[0].m_layer[:100,...].fft.transmission()
```

**Dlaczego nie działa:**
- `FFTTransmissionInterface` NIE otrzymuje `slice_info`
- `TransmissionCompute` używa `FFTCompute.load_data_from_zarr()` 
- `load_data_from_zarr()` zawsze ładuje pełny dataset: `data = data_set[...]`
- Slicing `:100` jest **ignorowany**, ładowane są WSZYSTKIE kroki czasowe!

---

## 🛠️ Obejście problemu (tymczasowe)

### Dla transmisji: użyj parametru `tmax`
```python
# Zamiast slicingu [:100]
trans_result = job[0].fft.transmission(tmax=100)
```

**Uwaga:** `tmax` działa INACZEJ niż slicing:
- `:100` oznacza "pierwsze 100 kroków z załadowanych danych"
- `tmax=100` oznacza "załaduj tylko pierwsze 100 kroków"

---

## 📊 Porównanie składni

| Funkcja | Obecna składnia | Status |
|---------|----------------|--------|
| **Dyspersja** | `job[0].m_layer[:100,...].fft.dispersion.compute_1d()` | ✅ Działa |
| **Transmisja** | `job[0].m_layer[:100,...].fft.transmission()` | ❌ Ignoruje slicing |
| **Transmisja (fix)** | `job[0].fft.transmission(tmax=100)` | ✅ Obejście |


### Parametry:
- **`x`** (float, wymagany): Pozycja x w metrach (np. `2000e-9` dla 2 µm)
  - Metoda automatycznie znajdzie najbliższą dostępną pozycję x
- **`freq_unit`** (str): Jednostka częstotliwości - "Hz", "kHz", "MHz", "GHz"
- **`trim_0f`** (int, opcjonalny): Liczba najniższych punktów do usunięcia
- **`flip`** (bool, opcjonalny): 🆕 Orientacja osi
  - `False` (domyślnie): Częstotliwość na osi X, Transmisja na osi Y
  - `True`: Częstotliwość na osi Y, Transmisja na osi X
  - **Użyj `flip=True` aby dopasować do wykresów dyspersji!**
- **`ax`** (Axes, opcjonalny): Matplotlib axes do rysowania
- **`**kwargs`**: Dodatkowe parametry matplotlib (color, linewidth, label, alpha, ...)

### Zwraca:
- `fig, ax` - Matplotlib figure i axes

### Przykłady:
```python
# Prosty przekrój (domyślna orientacja)

    linewidth=3,
```

# 🆕 NOWY PARAMETR: `flip` w `plot_transmission_crosssection()`

## Orientacja osi

**`flip=False`** (domyślnie):
- Częstotliwość na osi X (poziomo)

## 🔍 `spatial_window` vs `spatial_step` - Kluczowe różnice

### 📐 Definicje:

**`spatial_window`** = **Rozmiar okna analizy** (liczba punktów x w jednym oknie)
- Ile punktów przestrzennych używamy do **jednego** obliczenia FFT
- Przykład: `spatial_window=200` → każde FFT używa 200 sąsiednich punktów x
- **Wpływ na rozdzielczość częstotliwości:** `Δf = 1 / (window_size × Δx × dt)`
  - Większe okno → **lepsza rozdzielczość częstotliwości** (mniejsze Δf)
  - Mniejsze okno → **gorsza rozdzielczość częstotliwości** (większe Δf)

**`spatial_step`** = **Krok przesunięcia okna** (co ile punktów x tworzymy nowe okno)
- Jak często "przeskakujemy" do następnego okna
- Przykład: `spatial_step=1` → okno przesuwa się o 1 punkt (maksymalne nakładanie)
- Przykład: `spatial_step=200` → okno przesuwa się o cały swój rozmiar (zero nakładania)
- **Wpływ na liczbę okien:** `n_windows = (n_x - window_size) / step + 1`
  - Mniejszy step → **więcej okien** → lepsza rozdzielczość przestrzenna
  - Większy step → **mniej okien** → gorsza rozdzielczość przestrzenna (ale szybsze!)

---

### 🎯 Analogia: Mikroskop przesuwany po próbce

Wyobraź sobie mikroskop który przesuwa się wzdłuż próbki:

- **`spatial_window`** = **średnica pola widzenia mikroskopu**
  - Większa średnica → widzisz więcej jednocześnie → lepiej "uśredniasz" sygnał
  - Jak duży kawałek próbki analizujesz na raz

- **`spatial_step`** = **jak daleko przesuwasz mikroskop między kolejnymi pomiarami**
  - Mały krok (np. 1 punkt) → dużo nakładających się pomiarów → płynna mapa
  - Duży krok (np. cały rozmiar okna) → pomiary obok siebie bez nakładania

---

### 📊 Przykłady:

#### Przykład 1: Maksymalna rozdzielczość przestrzenna
```python
spatial_window=200,  # Okno 200 punktów
spatial_step=1       # Przesuwamy o 1 punkt
```
- **Liczba okien:** `n_windows ≈ n_x - 200` (prawie tyle co punktów!)
- **Nakładanie:** 199/200 = **99.5% nakładania**
- **Zastosowanie:** Gdy chcesz bardzo gładką mapę transmission(x, f)
- **Koszt:** Bardzo wolne! (3561 okien dla n_x=3761)

#### Przykład 2: Równowaga
```python
spatial_window=200,  # Okno 200 punktów
spatial_step=10      # Przesuwamy o 10 punktów
```
- **Liczba okien:** `n_windows ≈ (n_x - 200) / 10` = 356 okien
- **Nakładanie:** 190/200 = **95% nakładania**
- **Zastosowanie:** Dobry kompromis: gładka mapa + rozsądny czas
- **Koszt:** ~10× szybsze niż step=1

#### Przykład 3: Minimalne nakładanie
```python
spatial_window=200,   # Okno 200 punktów
spatial_step=200      # Przesuwamy o cały rozmiar okna
```
- **Liczba okien:** `n_windows = n_x / 200` ≈ 18 okien
- **Nakładanie:** **0% nakładania** (okna obok siebie)
- **Zastosowanie:** Gdy zależy Ci tylko na szybkości
- **Koszt:** Bardzo szybkie! (~200× szybsze niż step=1)

---

### ⚖️ Trade-offs:

| Parametr | Większa wartość | Mniejsza wartość |
|----------|----------------|------------------|
| **`spatial_window`** | ✅ Lepsza rozdzielczość częstotliwości<br>✅ Lepsze uśrednianie szumu<br>❌ Gorsza lokalizacja przestrzenna | ✅ Lepsza lokalizacja przestrzenna<br>❌ Gorsza rozdzielczość częstotliwości<br>❌ Więcej szumu |
| **`spatial_step`** | ✅ Szybsze obliczenia<br>✅ Mniej pamięci<br>❌ Gorsza rozdzielczość przestrzenna mapy | ✅ Lepsza rozdzielczość przestrzenna mapy<br>✅ Gładsze przejścia<br>❌ Wolniejsze obliczenia |

---

### 🧮 Matematyka:

Dla danych o wymiarach `(n_t, n_z, n_y, n_x, 3)`:

1. **Liczba okien:**
   ```python
   n_windows = (n_x - spatial_window) // spatial_step + 1
   ```
   - `n_x = 3761`, `window=200`, `step=1` → **3561 okien**
   - `n_x = 3761`, `window=200`, `step=10` → **357 okien**
   - `n_x = 3761`, `window=200`, `step=200` → **18 okien**

2. **Pozycje środków okien:**
   ```python
   window_centers = [window//2 + i*step for i in range(n_windows)]
   ```

3. **Procent nakładania:**
   ```python
   overlap = (spatial_window - spatial_step) / spatial_window * 100
   ```
   - `window=200, step=1` → **99.5% nakładania**
   - `window=200, step=100` → **50% nakładania**
   - `window=200, step=200` → **0% nakładania**

---

### 🎓 Kiedy używać:

**Duże okno + mały step** (`window=200, step=1`):
- Analizujesz **subtelne zmiany** transmission w przestrzeni
- Masz czas na długie obliczenia
- Chcesz **płynną mapę** bez skoków

**Średnie okno + średni step** (`window=200, step=10`):
- **Rekomendowane** dla większości przypadków
- Dobry kompromis rozdzielczość/czas

**Dowolne okno + step=okno** (`window=200, step=200`):
- Szybka **eksploracja danych**
- Nie zależy Ci na płynności mapy
- Okna są **niezależne** (zero nakładania)

### 🎓 Dlaczego spatial_window=1 powinno dać identyczny wynik?

**Matematyka:**

Dla `spatial_window=1, spatial_step=1`:

1. **Liczba okien:**
   ```
   n_windows = (n_x - 1) // 1 + 1 = n_x - 1 + 1 = n_x
   ```
   → Mamy **dokładnie n_x okien** (jedno dla każdego punktu x)

2. **Zawartość każdego okna:**
   ```
   Okno 0: data[:, x=0, mz]  → 1 punkt
   Okno 1: data[:, x=1, mz]  → 1 punkt
   ...
   Okno k: data[:, x=k, mz]  → 1 punkt
   ```

3. **FFT dla okna k:**
   ```python
   # W kodzie transmission:
   spectrum_k = FFT(data[:, x=k, mz])  # FFT po czasie
   power_k = mean(abs(spectrum_k)**2)   # mean po 1 punkcie = sam punkt
   
   # W prostej pętli:
   spectrum_k = FFT(data[:, x=k, mz])  # FFT po czasie
   power_k = abs(spectrum_k)**2
   ```

4. **Wynik:**
   ```
   power_map[:, k] = power_k  (dla każdego k = 0, 1, ..., n_x-1)
   ```
   
**Wniosek:** 
- Dla `window=1` nie ma uśredniania przestrzennego (tylko 1 punkt!)
- Każde okno = pojedynczy punkt x
- FFT dla okna k = FFT dla punktu x=k
- **Wyniki MUSZĄ być identyczne!**

---

**Jeśli wyniki się różnią, to błąd może być w:**
1. ❌ Indeksowaniu okien (np. błąd off-by-one)
2. ❌ Aplikowaniu window_function mimo window=1
3. ❌ Uśrednianiu po złej osi
4. ❌ Błędzie w liczbie okien

```python
        logger.warning(f"Using dt from zarr.attrs['{key}']: {self.dt} (dataset-specific not found)")
        break
```
